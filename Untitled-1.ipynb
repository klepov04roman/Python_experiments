```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Налаштування стилю для графіків
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# ==========================================
# PHASE 1: Quick Data Quality Check
# ==========================================
print("--- PHASE 1: Data Quality Check ---")

# Завантаження даних (замініть 'data.csv' на шлях до вашого файлу)
# Припускаємо наявність колонок: InvoiceNo, StockCode, Description, Quantity, InvoiceDate, UnitPrice, CustomerID, Country
try:
    df = pd.read_csv("OnlineRetail.csv", encoding="latin1")
except FileNotFoundError:
    # Створення фейкових даних для демонстрації, якщо файл не знайдено
    print("Файл 'OnlineRetail.csv' не знайдено. Створюємо тестовий набір даних...")
    np.random.seed(42)
    dates = pd.date_range(start="2025-01-01 08:00", periods=1000, freq="15min")
    df = pd.DataFrame(
        {
            "InvoiceNo": np.random.choice(
                ["536365", "C536379", "536380"], size=1000, p=[0.9, 0.08, 0.02]
            ),
            "StockCode": np.random.choice(
                ["85123A", "71053", "84406B", "22423"], size=1000
            ),
            "Description": np.random.choice(
                ["White Hanging Heart T-Light Holder", "White Metal Lantern", "Cream Cupid Hearts Coat Hanger", "Regency Cakestand 3 Tier"],
                size=1000,
            ),
            "Quantity": np.random.randint(-10, 50, size=1000),
            "InvoiceDate": np.random.choice(dates, size=1000),
            "UnitPrice": np.random.uniform(0.5, 12.5, size=1000),
            "CustomerID": np.random.choice(
                [17850.0, 13047.0, np.nan], size=1000, p=[0.4, 0.4, 0.2]
            ),
            "Country": np.random.choice(
                ["United Kingdom", "France", "Germany"], size=1000, p=[0.8, 0.1, 0.1]
            ),
        }
    )
    df["InvoiceDate"] = df["InvoiceDate"].astype(str)

# 1. Пошук негативних значень (повернення/скасування)
negative_qty = df[df["Quantity"] < 0]
print(f"Знайдено рядків з негативною кількістю (повернення): {len(negative_qty)}")

# 2. Перевірка пустих значень в CustomerID
null_customers = df["CustomerID"].isnull().sum()
print(f"Кількість замовлень без CustomerID (гостьові замовлення): {null_customers}")

# 3. Перевірка типів даних
print("\nТипи даних до очищення:")
print(df.dtypes)
print("-" * 40)


# ==========================================
# PHASE 2: Clean Your Data
# ==========================================
print("\n--- PHASE 2: Data Cleaning ---")

# Крок 1: Фільтрація даних
# Перетворюємо тип даних InvoiceNo на рядок для безпечного пошуку символу 'C'
df["InvoiceNo"] = df["InvoiceNo"].astype(str)

# Залишаємо тільки 'United Kingdom' та прибираємо повернення (InvoiceNo починається з 'C')
df_clean = df[
    (df["Country"] == "United Kingdom") & (~df["InvoiceNo"].str.startswith("C"))
].copy()

# Додатково прибираємо випадкові технічні негативні або нульові кількості товарів
df_clean = df_clean[df_clean["Quantity"] > 0]

# Конвертуємо дати у формат datetime
df_clean["InvoiceDate"] = pd.to_datetime(df_clean["InvoiceDate"])

# Крок 2: Розрахунок TotalSales
df_clean["TotalSales"] = df_clean["Quantity"] * df_clean["UnitPrice"]

print(f"Розмір очищеного датасету: {df_clean.shape[0]} рядків")
print("-" * 40)


# ==========================================
# PHASE 3: Run Your Analysis
# ==========================================
print("\n--- PHASE 3: Analysis ---")

# Аналіз 1: Top 10 Bestselling Products (за кількістю проданого)
top_10_quantity = (
    df_clean.groupby("Description")["Quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
print("\n[Топ-10 найпопулярніших товарів за кількістю]:")
print(top_10_quantity)

# Аналіз 2: Top 10 Revenue Products (за виторгом)
top_10_revenue = (
    df_clean.groupby("Description")["TotalSales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
print("\n[Топ-10 товарів, що принесли найбільший виторг]:")
print(top_10_revenue)

# Аналіз 3: Sales by Hour (Продажі за годинами)
df_clean["Hour"] = df_clean["InvoiceDate"].dt.hour
sales_by_hour = df_clean.groupby("Hour")["TotalSales"].sum()
print("\n[Продажі за годинами]:")
print(sales_by_hour)

# Аналіз 4: Sales by Day of Week (Продажі за днями тижня)
# .dt.day_name() повертає назву дня англійською
df_clean["DayOfWeek"] = df_clean["InvoiceDate"].dt.day_name()
# Впорядкуємо назви днів, щоб вони йшли логічно від понеділка до неділі
days_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday",
]
sales_by_day = (
    df_clean.groupby("DayOfWeek")["TotalSales"].sum().reindex(days_order)
)
print("\n[Продажі за днями тижня]:")
print(sales_by_day)
print("-" * 40)


# ==========================================
# PHASE 4: Visualise ONE Key Finding
# ==========================================
# Ми обираємо Option B: Продажі за годинами (лінійний графік)
print("\n--- PHASE 4: Visualisation ---")

plt.figure(figsize=(10, 5))
sns.lineplot(
    x=sales_by_hour.index,
    y=sales_by_hour.values,
    marker="o",
    linewidth=2.5,
    color="#2b7bba",
)

# Налаштування красивого візуалу
plt.title("Розподіл продажів за годинами в UK", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Година доби (Hour)", fontsize=11)
plt.ylabel("Загальний виторг (£)", fontsize=11)
plt.xticks(range(0, 24))  # Показуємо кожну годину на осі X
plt.grid(True, linestyle="--", alpha=0.6)

# Показуємо графік
plt.tight_layout()
plt.show()


# ==========================================
# PHASE 5: Write Your Insight
# ==========================================
print("\n--- PHASE 5: Insights & Recommendations ---")

# Автоматичний пошук пікової години для інсайту
peak_hour = sales_by_hour.idxmax()
total_revenue = sales_by_hour.sum()

# Розрахунок продажів у найактивніший проміжок (наприклад, з 10:00 до 15:00)
midday_sales = sales_by_hour.loc[10:15].sum()
percentage_midday = (midday_sales / total_revenue) * 100

print(f"--- ІНСАЙТ ---")
print(
    f"Найбільша купівельна активність спостерігається о {peak_hour}:00. \n"
    f"Близько {percentage_midday:.1f}% від усього денного виторгу припадає на прайм-тайм з 10:00 до 15:00."
)
print()
print(f"--- РЕКОМЕНДАЦІЯ ---")
print(
    f"Рекомендується планувати запуск маркетингових розсилок та push-повідомлень на 09:30 - 11:30, \n"
    f"щоб підігріти інтерес користувачів безпосередньо перед настанням щоденного піку продажів."
)
```
